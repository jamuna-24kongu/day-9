```python
import pandas as pd
import numpy as np

# ============================================================
# COMPANY EMPLOYEE DATASET - COMPLETE DATA CLEANING
# ============================================================

# 1. LOAD DATASET
# The 'FileNotFoundError' occurred because the dataset was not found at the specified path.
# Please ensure the 'Day11_Messy_Company_Employee_Dataset.csv' file is uploaded to your Colab environment.
# If you upload it directly, it will likely be in '/content/'.
file_path = "/content/Day11_Messy_Company_Employee_Dataset.csv"
df = pd.read_csv(file_path)

# Keep original dataset for comparison
original_df = df.copy()

print("========== ORIGINAL DATASET ==========")
print(df.head())

print("\nShape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types:")
print(df.dtypes)

# 2. DATASET INFORMATION
print("\n========== DATASET INFORMATION ==========")
df.info()

# 3. STATISTICAL SUMMARY
print("\n========== STATISTICAL SUMMARY ==========")
print(df.describe(include="all"))

# 4. MISSING VALUES
print("\n========== MISSING VALUES BEFORE CLEANING ==========")
print(df.isnull().sum())
print("Total Missing Values:", df.isnull().sum().sum())

# 5. DUPLICATES
print("\n========== DUPLICATES BEFORE CLEANING ==========")
print("Duplicate Rows:", df.duplicated().sum())

# 6. CLEAN COLUMN NAMES
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# 7. CLEAN TEXT VALUES
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype("string").str.strip()

# 8. STANDARDIZE MISSING VALUES
missing_values = [
    "", " ", "NA", "N/A", "na", "n/a",
    "NULL", "null", "None", "none", "?", "-"
]

df = df.replace(missing_values, np.nan)

# 9. CONVERT NUMERIC COLUMNS
for col in df.columns:
    converted = pd.to_numeric(df[col], errors="coerce")

    if df[col].notna().sum() > 0:
        ratio = converted.notna().sum() / df[col].notna().sum()

        if ratio >= 0.8:
            df[col] = converted

# 10. CONVERT DATE COLUMN
for col in df.columns:
    if "date" in col.lower() or "joining" in col.lower():
        df[col] = pd.to_datetime(df[col], errors="coerce")

# 11. HANDLE NUMERICAL MISSING VALUES USING MEDIAN
print("\n========== NUMERICAL IMPUTATION ==========")

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    missing = df[col].isnull().sum()

    if missing > 0:
        median_value = df[col].median()
        df[col] = df[col].fillna(median_value)

        print(
            f"{col}: {missing} missing values "
            f"filled with median = {median_value}"
        )

# 12. HANDLE CATEGORICAL MISSING VALUES USING MODE
print("\n========== CATEGORICAL IMPUTATION ==========")

categorical_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for col in categorical_columns:
    missing = df[col].isnull().sum()

    if missing > 0:
        mode_value = df[col].mode()

        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value.iloc[0])

            print(
                f"{col}: {missing} missing values "
                f"filled with mode = {mode_value.iloc[0]}"
            )

# 13. FORWARD FILL REMAINING MISSING VALUES
df = df.ffill()

# 14. DROP REMAINING MISSING ROWS IF ANY
if df.isnull().sum().sum() > 0:
    df = df.dropna()

# 15. STANDARDIZE TEXT ENTRIES
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].str.strip().str.title()

# 16. REMOVE DUPLICATES
duplicates = df.duplicated().sum()

df = df.drop_duplicates()

print("\n========== DUPLICATE CLEANING ==========")
print("Duplicates Removed:", duplicates)
print("Duplicates Remaining:", df.duplicated().sum())

# 17. FINAL DATASET
print("\n========== CLEANED DATASET ==========")
print(df.head())

print("\nCleaned Shape:", df.shape)

# 18. FINAL DATA TYPES
print("\n========== DATA TYPES AFTER CLEANING ==========")
print(df.dtypes)

# 19. FINAL MISSING VALUES
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

print(
    "Total Missing Values After Cleaning:",
    df.isnull().sum().sum()
)

# 20. FINAL DUPLICATES
print("\n========== DUPLICATES AFTER CLEANING ==========")
print("Duplicate Rows:", df.duplicated().sum())

# 21. BEFORE VS AFTER COMPARISON
print("\n========== BEFORE VS AFTER CLEANING ==========")

print("Rows:")
print("Before:", original_df.shape[0])
print("After :", df.shape[0])

print("\nColumns:")
print("Before:", original_df.shape[1])
print("After :", df.shape[1])

print("\nMissing Values:")
print("Before:", original_df.isnull().sum().sum())
print("After :", df.isnull().sum().sum())

print("\nDuplicate Rows:")
print("Before:", original_df.duplicated().sum())
print("After :", df.duplicated().sum())

# 22. FINAL STATISTICS
print("\n========== FINAL STATISTICAL SUMMARY ==========")
print(df.describe(include="all"))

# 23. EXPORT CLEANED DATASET
output_file = "Cleaned_Company_Employee_Dataset.csv"

df.to_csv(output_file, index=False)

print("\n========== EXPORT ==========")
print("Cleaned dataset saved as:", output_file)

# 24. CLEANING SUMMARY
print("\n========== CLEANING SUMMARY ==========")

print("""
✓ Dataset loaded using Pandas
✓ Dataset structure and statistics inspected
✓ Missing values identified using isnull()
✓ Missing numerical values filled using median
✓ Missing categorical values filled using mode
✓ Forward filling applied where required
✓ Remaining missing rows removed using dropna()
✓ Column names standardized
✓ Text values cleaned and standardized
✓ Numeric columns converted to appropriate datatype
✓ Joining date converted to datetime
✓ Duplicate records removed using drop_duplicates()
✓ Dataset verified after cleaning
✓ Before-and-after comparison performed
✓ Cleaned dataset exported as CSV
""")
```
