```python
# ================================================================
# DAY 13 - RESTAURANT BRANCH PERFORMANCE DATASET
# COMPLETE EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================

# =========================
# 1. IMPORT LIBRARIES
# =========================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("=" * 70)
print("RESTAURANT BRANCH PERFORMANCE - EXPLORATORY DATA ANALYSIS")
print("=" * 70)


# =========================
# 2. LOAD DATASET
# =========================

file_path = "Day13_Restaurant_Branch_Performance_Dataset.csv"

df = pd.read_csv(file_path)

print("\nDATASET LOADED SUCCESSFULLY")
print("Shape:", df.shape)


# =========================
# 3. BASIC DATASET INSPECTION
# =========================

print("\n" + "=" * 70)
print("FIRST 5 ROWS")
print("=" * 70)
print(df.head())


print("\n" + "=" * 70)
print("LAST 5 ROWS")
print("=" * 70)
print(df.tail())


print("\n" + "=" * 70)
print("RANDOM SAMPLE")
print("=" * 70)
print(df.sample(min(5, len(df)), random_state=42))


print("\n" + "=" * 70)
print("DATASET SHAPE")
print("=" * 70)
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])


print("\n" + "=" * 70)
print("COLUMN NAMES")
print("=" * 70)
print(df.columns.tolist())


print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)
print(df.dtypes)


print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)
df.info()


# =========================
# 4. MISSING VALUES
# =========================

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

missing = df.isnull().sum()

print(missing)

print("\nTotal missing values:", missing.sum())


# =========================
# 5. DUPLICATE VALUES
# =========================

print("\n" + "=" * 70)
print("DUPLICATE RECORDS")
print("=" * 70)

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)


# =========================
# 6. UNIQUE VALUES
# =========================

print("\n" + "=" * 70)
print("UNIQUE VALUES")
print("=" * 70)

for col in df.select_dtypes(include="object").columns:
    print("\n", col)
    print(df[col].nunique(), "unique values")
    print(df[col].unique())


# =========================
# 7. DESCRIPTIVE STATISTICS
# =========================

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

print(df.describe())


print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS INCLUDING CATEGORICAL DATA")
print("=" * 70)

print(df.describe(include="all"))


# =========================
# 8. SEPARATE NUMERICAL
#    AND CATEGORICAL COLUMNS
# =========================

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()

print("\n" + "=" * 70)
print("NUMERICAL COLUMNS")
print("=" * 70)

print(numerical_columns)


print("\n" + "=" * 70)
print("CATEGORICAL COLUMNS")
print("=" * 70)

print(categorical_columns)


# =========================
# 9. NUMERICAL SUMMARY
# =========================

print("\n" + "=" * 70)
print("NUMERICAL COLUMN SUMMARY")
print("=" * 70)

for col in numerical_columns:

    print("\nColumn:", col)
    print("Mean   :", round(df[col].mean(), 2))
    print("Median :", round(df[col].median(), 2))
    print("Minimum:", round(df[col].min(), 2))
    print("Maximum:", round(df[col].max(), 2))
    print("Std Dev:", round(df[col].std(), 2))


# =========================
# 10. CATEGORICAL SUMMARY
# =========================

print("\n" + "=" * 70)
print("CATEGORICAL COLUMN SUMMARY")
print("=" * 70)

for col in categorical_columns:

    print("\nColumn:", col)
    print(df[col].value_counts())


# =========================
# 11. DISTRIBUTION ANALYSIS
# =========================

print("\n" + "=" * 70)
print("DISTRIBUTION ANALYSIS")
print("=" * 70)

for col in numerical_columns:

    plt.figure(figsize=(8, 5))

    sns.histplot(
        df[col].dropna(),
        kde=True
    )

    plt.title("Distribution of " + col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()


# =========================
# 12. BOXPLOT ANALYSIS
# =========================

for col in numerical_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(
        x=df[col]
    )

    plt.title("Boxplot of " + col)
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()


# =========================
# 13. CATEGORICAL DISTRIBUTIONS
# =========================

for col in categorical_columns:

    # Avoid plotting extremely high-cardinality columns
    if df[col].nunique() <= 15:

        plt.figure(figsize=(9, 5))

        order = df[col].value_counts().index

        sns.countplot(
            data=df,
            x=col,
            order=order
        )

        plt.title("Distribution of " + col)
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()


# =========================
# 14. CORRELATION MATRIX
# =========================

print("\n" + "=" * 70)
print("CORRELATION MATRIX")
print("=" * 70)

correlation_matrix = df[numerical_columns].corr()

print(correlation_matrix.round(2))


# =========================
# 15. CORRELATION HEATMAP
# =========================

plt.figure(
    figsize=(12, 8)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.show()


# =========================
# 16. STRONGEST CORRELATIONS
# =========================

print("\n" + "=" * 70)
print("STRONGEST CORRELATION PAIRS")
print("=" * 70)

corr_pairs = (
    correlation_matrix
    .where(
        np.triu(
            np.ones(correlation_matrix.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .sort_values(
        key=abs,
        ascending=False
    )
)

print(corr_pairs.head(10))


# =========================
# 17. BRANCH PERFORMANCE
# =========================

# Find likely branch column automatically
branch_col = None

for col in df.columns:
    if "branch" in col.lower():
        branch_col = col
        break

if branch_col:

    print("\n" + "=" * 70)
    print("PERFORMANCE BY BRANCH")
    print("=" * 70)

    branch_summary = df.groupby(branch_col)[
        numerical_columns
    ].mean().round(2)

    print(branch_summary)

    # Revenue comparison
    revenue_cols = [
        col for col in numerical_columns
        if "revenue" in col.lower()
    ]

    if revenue_cols:

        revenue_col = revenue_cols[0]

        branch_revenue = (
            df.groupby(branch_col)[revenue_col]
            .mean()
            .sort_values(ascending=False)
        )

        print("\nAverage Revenue by Branch:")
        print(branch_revenue)

        plt.figure(figsize=(10, 5))

        branch_revenue.plot(
            kind="bar"
        )

        plt.title("Average Revenue by Branch")
        plt.xlabel("Branch")
        plt.ylabel("Average Revenue")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()


# =========================
# 18. REGION PERFORMANCE
# =========================

region_col = None

for col in df.columns:
    if "region" in col.lower():
        region_col = col
        break

if region_col:

    print("\n" + "=" * 70)
    print("PERFORMANCE BY REGION")
    print("=" * 70)

    region_summary = df.groupby(region_col)[
        numerical_columns
    ].mean().round(2)

    print(region_summary)

    if revenue_cols:

        region_revenue = (
            df.groupby(region_col)[revenue_col]
            .mean()
            .sort_values(ascending=False)
        )

        print("\nAverage Revenue by Region:")
        print(region_revenue)

        plt.figure(figsize=(9, 5))

        region_revenue.plot(
            kind="bar"
        )

        plt.title("Average Revenue by Region")
        plt.xlabel("Region")
        plt.ylabel("Average Revenue")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()


# =========================
# 19. STORE TYPE PERFORMANCE
# =========================

store_type_col = None

for col in df.columns:
    if "store" in col.lower() and "type" in col.lower():
        store_type_col = col
        break

if store_type_col:

    print("\n" + "=" * 70)
    print("PERFORMANCE BY STORE TYPE")
    print("=" * 70)

    store_summary = df.groupby(store_type_col)[
        numerical_columns
    ].mean().round(2)

    print(store_summary)

    if revenue_cols:

        store_revenue = (
            df.groupby(store_type_col)[revenue_col]
            .mean()
            .sort_values(ascending=False)
        )

        print("\nAverage Revenue by Store Type:")
        print(store_revenue)

        plt.figure(figsize=(8, 5))

        store_revenue.plot(
            kind="bar"
        )

        plt.title("Average Revenue by Store Type")
        plt.xlabel("Store Type")
        plt.ylabel("Average Revenue")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()


# =========================
# 20. IMPORTANT BUSINESS
#     FACTORS
# =========================

print("\n" + "=" * 70)
print("IMPORTANT BUSINESS FACTORS")
print("=" * 70)

important_keywords = [
    "customer",
    "revenue",
    "profit",
    "marketing",
    "staff",
    "delivery",
    "rating"
]

important_columns = []

for col in numerical_columns:

    if any(
        keyword in col.lower()
        for keyword in important_keywords
    ):
        important_columns.append(col)

print(
    "Important numerical columns found:"
)

print(important_columns)


# =========================
# 21. RELATIONSHIP WITH REVENUE
# =========================

if revenue_cols:

    print("\n" + "=" * 70)
    print("CORRELATION WITH REVENUE")
    print("=" * 70)

    revenue_correlations = (
        correlation_matrix[revenue_col]
        .drop(revenue_col)
        .sort_values(
            key=abs,
            ascending=False
        )
    )

    print(revenue_correlations)


# =========================
# 22. RELATIONSHIP WITH PROFIT
# =========================

profit_cols = [
    col for col in numerical_columns
    if "profit" in col.lower()
]

if profit_cols:

    profit_col = profit_cols[0]

    print("\n" + "=" * 70)
    print("CORRELATION WITH PROFIT")
    print("=" * 70)

    profit_correlations = (
        correlation_matrix[profit_col]
        .drop(profit_col)
        .sort_values(
            key=abs,
            ascending=False
        )
    )

    print(profit_correlations)


# =========================
# 23. SCATTER PLOTS FOR
#     IMPORTANT RELATIONSHIPS
# =========================

if revenue_cols:

    for col in important_columns:

        if col != revenue_col:

            plt.figure(figsize=(8, 5))

            sns.scatterplot(
                data=df,
                x=col,
                y=revenue_col
            )

            plt.title(
                f"{col} vs {revenue_col}"
            )

            plt.xlabel(col)
            plt.ylabel(revenue_col)

            plt.tight_layout()
            plt.show()


# =========================
# 24. TOP AND BOTTOM
#     PERFORMING RECORDS
# =========================

if revenue_cols:

    print("\n" + "=" * 70)
    print("TOP 10 RECORDS BY REVENUE")
    print("=" * 70)

    print(
        df.nlargest(
            10,
            revenue_col
        )
    )


    print("\n" + "=" * 70)
    print("BOTTOM 10 RECORDS BY REVENUE")
    print("=" * 70)

    print(
        df.nsmallest(
            10,
            revenue_col
        )
    )


# =========================
# 25. AUTOMATIC DATA-DRIVEN
#     OBSERVATIONS
# =========================

print("\n" + "=" * 70)
print("DATA-DRIVEN OBSERVATIONS")
print("=" * 70)

observations = []

# Observation 1 - Dataset size
observations.append(
    f"1. The dataset contains {df.shape[0]} records "
    f"and {df.shape[1]} variables, providing a broad view "
    f"of restaurant branch performance."
)

# Observation 2 - Missing values
total_missing = df.isnull().sum().sum()

if total_missing == 0:

    observations.append(
        "2. The dataset contains no missing values, "
        "so the available observations can be used directly "
        "for exploratory analysis."
    )

else:

    observations.append(
        f"2. The dataset contains {total_missing} missing values "
        "that should be considered when interpreting the results."
    )


# Observation 3 - Revenue
if revenue_cols:

    max_rev = df[revenue_col].max()
    min_rev = df[revenue_col].min()
    avg_rev = df[revenue_col].mean()

    observations.append(
        f"3. Average {revenue_col} is {avg_rev:.2f}, "
        f"with values ranging from {min_rev:.2f} "
        f"to {max_rev:.2f}, indicating variation "
        "in revenue performance across observations."
    )


# Observation 4 - Branch
if branch_col and revenue_cols:

    best_branch = (
        df.groupby(branch_col)[revenue_col]
        .mean()
        .idxmax()
    )

    best_branch_value = (
        df.groupby(branch_col)[revenue_col]
        .mean()
        .max()
    )

    observations.append(
        f"4. {best_branch} has the highest average "
        f"{revenue_col} at approximately "
        f"{best_branch_value:.2f}, making it the strongest "
        "performing branch based on average revenue."
    )


# Observation 5 - Region
if region_col and revenue_cols:

    best_region = (
        df.groupby(region_col)[revenue_col]
        .mean()
        .idxmax()
    )

    observations.append(
        f"5. The {best_region} region has the highest "
        f"average {revenue_col}, indicating stronger "
        "revenue performance compared with other regions."
    )


# Observation 6 - Store type
if store_type_col and revenue_cols:

    best_store = (
        df.groupby(store_type_col)[revenue_col]
        .mean()
        .idxmax()
    )

    observations.append(
        f"6. The {best_store} store type records the highest "
        f"average {revenue_col}, suggesting that store format "
        "may influence restaurant performance."
    )


# Observation 7 - Strongest correlation
if len(corr_pairs) > 0:

    pair = corr_pairs.index[0]
    value = corr_pairs.iloc[0]

    observations.append(
        f"7. The strongest numerical relationship is between "
        f"{pair[0]} and {pair[1]}, with a correlation coefficient "
        f"of {value:.2f}. This indicates a "
        f"{'positive' if value > 0 else 'negative'} relationship."
    )


# Observation 8 - Profit
if profit_cols:

    avg_profit = df[profit_col].mean()

    observations.append(
        f"8. Average {profit_col} is {avg_profit:.2f}. "
        "Differences in profit across branches can help "
        "identify locations with stronger operational efficiency."
    )


# Print only first 8 observations
for observation in observations[:8]:

    print("\n" + observation)


# =========================
# 26. FINAL SUMMARY
# =========================

print("\n" + "=" * 70)
print("EDA COMPLETED")
print("=" * 70)

print("""
The Exploratory Data Analysis examined:

1. Dataset structure and data types
2. Missing values and duplicate records
3. Descriptive statistics
4. Numerical and categorical distributions
5. Revenue and profit performance
6. Customer and operational factors
7. Branch-wise performance
8. Region-wise performance
9. Store-type performance
10. Correlation between numerical variables
11. Important relationships using scatter plots
12. Data-driven business observations

EDA PROCESS COMPLETED SUCCESSFULLY.
""")
```
