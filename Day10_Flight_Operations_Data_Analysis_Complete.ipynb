# Flight Operations Data Analysis Using Pandas

## Objective
Perform a complete data analysis workflow on the Flight Operations dataset using Python and Pandas. The analysis includes loading the dataset, understanding its structure, descriptive statistics, data selection, filtering, sorting, grouping, aggregation, transformation, and identifying key findings.

> **Note:** Upload the Flight Operations CSV when prompted. The notebook uses the dataset's actual columns dynamically where possible.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

# LMS Flight Operations dataset
try:
    df = pd.read_csv("/content/Day10_Flight_Operations_Dataset.csv")
except FileNotFoundError:
    uploaded = files.upload()
    filename = list(uploaded.keys())[0]
    df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

## 1. Inspect the Dataset

```python
print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nRandom 5 rows:")
display(df.sample(min(5, len(df))))
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nIndex:")
print(df.index)

print("\nData types:")
display(df.dtypes)
```

## Flight Operations Dataset — Column Overview

The dataset contains 180 flight-operation records and 17 columns covering flight details, passenger information, pricing, delays, status, weather, booking channel, baggage, meals, and passenger satisfaction.


```python
print("Dataset information:")
df.info()
```

```python
print("Descriptive statistics:")
display(df.describe(include="all"))
```

## 2. Data Quality Checks

```python
print("Missing values:")
display(df.isnull().sum())

print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates().copy()

print("Shape after removing duplicates:", df.shape)
```

```python
# Display unique-value counts for each column
for column in df.columns:
    print(f"{column}: {df[column].nunique()} unique values")
```

## 3. Data Selection

```python
print("Available columns:")
print(df.columns.tolist())

# Select the first column
print("\nFirst column:")
display(df[[df.columns[0]]].head(10))

# Select the first three columns
print("\nFirst three columns:")
display(df.iloc[:, :3].head(10))
```

## 4. Filtering

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(include="object").columns.tolist()

print("Numeric columns:", numeric_columns)
print("Categorical columns:", categorical_columns)

if numeric_columns:
    column = numeric_columns[0]
    median_value = df[column].median()

    print(f"\nRecords where {column} is greater than its median ({median_value:.2f}):")
    display(df[df[column] > median_value].head(10))
else:
    print("No numeric columns available for numeric filtering.")
```

## 5. Sorting

```python
if numeric_columns:
    column = numeric_columns[0]

    print(f"Top 10 records sorted by {column}:")
    display(df.sort_values(by=column, ascending=False).head(10))

    print(f"Bottom 10 records sorted by {column}:")
    display(df.sort_values(by=column, ascending=True).head(10))
else:
    print("No numeric columns available for sorting.")
```

## 6. Numerical Analysis

```python
if numeric_columns:
    summary = pd.DataFrame({
        "Minimum": df[numeric_columns].min(),
        "Maximum": df[numeric_columns].max(),
        "Mean": df[numeric_columns].mean(),
        "Median": df[numeric_columns].median(),
        "Standard Deviation": df[numeric_columns].std()
    })
    display(summary)
else:
    print("No numeric columns available.")
```

## 7. Grouping and Aggregation

```python
if categorical_columns and numeric_columns:
    category = categorical_columns[0]
    value = numeric_columns[0]

    grouped = (
        df.groupby(category)[value]
          .agg(["count", "mean", "min", "max", "sum"])
          .sort_values("count", ascending=False)
    )

    print(f"Grouped analysis: {category} vs {value}")
    display(grouped.head(15))
else:
    print("Suitable categorical/numeric columns were not found.")
```

```python
if categorical_columns:
    category = categorical_columns[0]

    print(f"Frequency of {category}:")
    display(df[category].value_counts().head(15))
else:
    print("No categorical columns available.")
```

## 8. Data Transformation

```python
if numeric_columns:
    column = numeric_columns[0]

    # Min-max normalization
    min_value = df[column].min()
    max_value = df[column].max()

    if max_value != min_value:
        df[column + "_Normalized"] = (
            (df[column] - min_value) / (max_value - min_value)
        )
    else:
        df[column + "_Normalized"] = 0

    # Divide values into Low, Medium, and High categories
    df[column + "_Category"] = pd.cut(
        df[column],
        bins=3,
        labels=["Low", "Medium", "High"],
        include_lowest=True
    )

    display(df.head(10))
else:
    print("No numeric column available for transformation.")
```

```python
if numeric_columns:
    category_column = numeric_columns[0] + "_Category"

    print("Transformed category distribution:")
    display(df[category_column].value_counts().sort_index())
```

## 9. Correlation Analysis

```python
numeric_df = df.select_dtypes(include=np.number)

if numeric_df.shape[1] > 1:
    display(numeric_df.corr())
else:
    print("At least two numeric columns are required for correlation analysis.")
```

## 10. Visual Analysis

```python
if numeric_columns:
    column = numeric_columns[0]

    plt.figure(figsize=(10, 5))
    plt.hist(df[column].dropna(), bins=20)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"Distribution of {column}")
    plt.show()
else:
    print("No numeric columns available.")
```

```python
if categorical_columns:
    category = categorical_columns[0]
    counts = df[category].value_counts().head(10)

    plt.figure(figsize=(10, 5))
    counts.plot(kind="bar")
    plt.xlabel(category)
    plt.ylabel("Number of Records")
    plt.title(f"Top 10 {category} Categories")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print("No categorical columns available.")
```

## 13. Flight-Specific Analysis Questions

We use Pandas to answer meaningful questions about the flight operations:

- Which airline operates the most flights?
- Which origin and destination are most common?
- Which flight status occurs most often?
- What is the average number of passengers per flight?
- Which flights have the highest delays?
- Which airline has the highest average passenger satisfaction?
- Which booking channel is most commonly used?
- How does average ticket price vary by travel class?


```python
# Airline-wise flight count
airline_flights = (
    df.groupby("Airline")["Flight_ID"]
      .count()
      .sort_values(ascending=False)
)

display(airline_flights.to_frame("Flight_Count"))
```

```python
# Origin and destination frequency
print("Top origins:")
display(df["Origin"].value_counts().head(10).to_frame("Flight_Count"))

print("Top destinations:")
display(df["Destination"].value_counts().head(10).to_frame("Flight_Count"))
```

```python
# Flight status analysis
status_analysis = df["Flight_Status"].value_counts().to_frame("Flight_Count")
display(status_analysis)
```

```python
# Passenger and capacity analysis
passenger_summary = df[[
    "Passengers",
    "Seat_Capacity",
    "Average_Ticket_Price",
    "Delay_Minutes",
    "Avg_Baggage_Kg",
    "Passenger_Satisfaction"
]].agg(["count", "mean", "min", "max", "median"])

display(passenger_summary)
```

```python
# Top 10 most delayed flights
top_delayed = df.sort_values(
    by="Delay_Minutes",
    ascending=False
)[[
    "Flight_ID", "Airline", "Origin", "Destination",
    "Delay_Minutes", "Flight_Status", "Weather"
]].head(10)

display(top_delayed)
```

```python
# Average passenger satisfaction by airline
airline_satisfaction = (
    df.groupby("Airline")["Passenger_Satisfaction"]
      .agg(["count", "mean", "min", "max"])
      .sort_values("mean", ascending=False)
)

display(airline_satisfaction)
```

```python
# Booking channel usage
booking_analysis = (
    df["Booking_Channel"]
      .value_counts()
      .to_frame("Booking_Count")
)

display(booking_analysis)
```

```python
# Average ticket price by travel class
class_price = (
    df.groupby("Travel_Class")["Average_Ticket_Price"]
      .agg(["count", "mean", "min", "max"])
      .sort_values("mean", ascending=False)
)

display(class_price)
```

```python
# Airline performance summary
airline_summary = (
    df.groupby("Airline")
      .agg(
          Flights=("Flight_ID", "count"),
          Avg_Passengers=("Passengers", "mean"),
          Avg_Ticket_Price=("Average_Ticket_Price", "mean"),
          Avg_Delay=("Delay_Minutes", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean")
      )
      .sort_values("Flights", ascending=False)
)

display(airline_summary)
```

```python
# Create useful transformed columns
df["Seat_Utilization_Pct"] = (
    df["Passengers"] / df["Seat_Capacity"] * 100
).round(2)

df["Delay_Category"] = pd.cut(
    df["Delay_Minutes"],
    bins=[-1, 0, 15, 60, np.inf],
    labels=["On Time", "Minor Delay", "Moderate Delay", "Major Delay"]
)

display(df[[
    "Flight_ID",
    "Passengers",
    "Seat_Capacity",
    "Seat_Utilization_Pct",
    "Delay_Minutes",
    "Delay_Category"
]].head(10))
```

```python
# Analyze transformed columns
print("Delay category distribution:")
display(df["Delay_Category"].value_counts().sort_index())

print("Average seat utilization by airline:")
display(
    df.groupby("Airline")["Seat_Utilization_Pct"]
      .mean()
      .sort_values(ascending=False)
      .to_frame("Avg_Seat_Utilization_Pct")
)
```

```python
# Weather and delay analysis
weather_delay = (
    df.groupby("Weather")["Delay_Minutes"]
      .agg(["count", "mean", "max"])
      .sort_values("mean", ascending=False)
)

display(weather_delay)
```

```python
# Travel class and satisfaction
class_satisfaction = (
    df.groupby("Travel_Class")["Passenger_Satisfaction"]
      .agg(["count", "mean", "min", "max"])
      .sort_values("mean", ascending=False)
)

display(class_satisfaction)
```
