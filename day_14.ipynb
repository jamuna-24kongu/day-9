```python
# ============================================================
# DAY 14 - FOOD DELIVERY BUSINESS PERFORMANCE
# Visualization Portfolio using Matplotlib and Seaborn
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Plot settings
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# -----------------------------
# 2. LOAD DATASET
# -----------------------------
# Original path: file_path = "/mnt/data/Day14_Food_Delivery_Visualization_Dataset.csv"
# Updated path for typical Colab upload to the session root or /content/
file_path = "Day14_Food_Delivery_Visualization_Dataset.csv"

df = pd.read_csv(file_path)

print("=" * 70)
print("FOOD DELIVERY BUSINESS PERFORMANCE DATASET")
print("=" * 70)

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())


# ============================================================
# 3. DATA PREPARATION
# ============================================================

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Sort data by date
df = df.sort_values("Date")

# Create numerical month order
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

# Display descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())


# ============================================================
# 4. LINE PLOT - ORDERS OVER TIME
# ============================================================

daily_orders = df.groupby("Date")["Orders"].sum().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(
    data=daily_orders,
    x="Date",
    y="Orders",
    marker="o"
)

plt.title("Food Delivery Orders Over Time", fontsize=16, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The line plot shows how food delivery orders change over time.
The number of orders fluctuates across different dates, indicating
variation in customer demand. This can help the business identify
periods of high and low order activity and plan staffing and
delivery resources accordingly.
""")


# ============================================================
# 5. LINE PLOT - REVENUE OVER TIME
# ============================================================

daily_revenue = df.groupby("Date")["Revenue"].sum().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(
    data=daily_revenue,
    x="Date",
    y="Revenue",
    marker="o"
)

plt.title("Food Delivery Revenue Over Time", fontsize=16, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
Revenue varies over time along with changes in business activity.
Higher-revenue periods indicate stronger business performance.
The company can use these patterns to plan promotions, inventory,
and operational resources during high-demand periods.
""")


# ============================================================
# 6. BAR CHART - REVENUE BY CITY
# ============================================================

city_revenue = (
    df.groupby("City")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(11, 6))
sns.barplot(
    data=city_revenue,
    x="City",
    y="Revenue"
)

plt.title("Total Revenue by City", fontsize=16, fontweight="bold")
plt.xlabel("City")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nINTERPRETATION:")

top_city = city_revenue.iloc[0]
low_city = city_revenue.iloc[-1]

print(
    f"{top_city['City']} generates the highest total revenue, "
    f"while {low_city['City']} generates the lowest among the cities."
)
print("""
This indicates that business performance differs across locations.
The company can focus expansion and marketing strategies on
high-performing cities while investigating opportunities in
lower-performing locations.
""")


# ============================================================
# 7. BAR CHART - ORDERS BY CUISINE
# ============================================================

cuisine_orders = (
    df.groupby("Cuisine")["Orders"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=cuisine_orders,
    x="Cuisine",
    y="Orders"
)

plt.title("Total Orders by Cuisine", fontsize=16, fontweight="bold")
plt.xlabel("Cuisine")
plt.ylabel("Total Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nINTERPRETATION:")

top_cuisine = cuisine_orders.iloc[0]

print(
    f"{top_cuisine['Cuisine']} has the highest total number of orders."
)
print("""
The chart shows differences in customer preference across cuisines.
Popular cuisines can receive greater promotional attention, while
less popular cuisines may require targeted offers or menu improvements.
""")


# ============================================================
# 8. BAR CHART - REVENUE BY CUISINE
# ============================================================

cuisine_revenue = (
    df.groupby("Cuisine")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=cuisine_revenue,
    x="Cuisine",
    y="Revenue"
)

plt.title("Total Revenue by Cuisine", fontsize=16, fontweight="bold")
plt.xlabel("Cuisine")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nINTERPRETATION:")

top_revenue_cuisine = cuisine_revenue.iloc[0]

print(
    f"{top_revenue_cuisine['Cuisine']} generates the highest total revenue."
)
print("""
Revenue differences between cuisines are not necessarily determined
only by order volume. Average order value and customer purchasing
patterns can also influence revenue.
""")


# ============================================================
# 9. SCATTER PLOT - MARKETING SPEND VS REVENUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Marketing_Spend",
    y="Revenue",
    hue="City",
    s=80
)

plt.title(
    "Relationship Between Marketing Spend and Revenue",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")
plt.legend(title="City", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()

# Correlation
marketing_corr = df["Marketing_Spend"].corr(df["Revenue"])

print("\nINTERPRETATION:")
print(
    f"The correlation between marketing spend and revenue is "
    f"{marketing_corr:.2f}."
)

print("""
The scatter plot helps examine whether higher marketing spending
is associated with higher revenue. A positive relationship suggests
that marketing investment may be associated with improved revenue,
although correlation alone does not prove causation.
""")


# ============================================================
# 10. SCATTER PLOT - DELIVERY TIME VS CUSTOMER RATING
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Avg_Delivery_Minutes",
    y="Customer_Rating",
    hue="Weather",
    s=80
)

plt.title(
    "Relationship Between Delivery Time and Customer Rating",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Average Delivery Time (Minutes)")
plt.ylabel("Customer Rating")
plt.legend(title="Weather", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()

delivery_rating_corr = df["Avg_Delivery_Minutes"].corr(
    df["Customer_Rating"]
)

print("\nINTERPRETATION:")
print(
    f"The correlation between delivery time and customer rating is "
    f"{delivery_rating_corr:.2f}."
)

print("""
The chart shows a strong negative relationship between delivery
time and customer rating. As delivery time increases, customer
ratings tend to decrease. This suggests that improving delivery
speed could contribute to better customer satisfaction.
""")


# ============================================================
# 11. HISTOGRAM - DISTRIBUTION OF ORDERS
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Orders",
    bins=20,
    kde=True
)

plt.title("Distribution of Food Delivery Orders", fontsize=16, fontweight="bold")
plt.xlabel("Number of Orders")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The histogram shows the distribution of order volumes.
Most observations are concentrated around the central range,
while relatively high or low order volumes occur less frequently.
This helps understand normal business demand and identify unusual
order periods.
""")


# ============================================================
# 12. HISTOGRAM - CUSTOMER RATINGS
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Customer_Rating",
    bins=15,
    kde=True
)

plt.title("Distribution of Customer Ratings", fontsize=16, fontweight="bold")
plt.xlabel("Customer Rating")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
Customer ratings are concentrated toward the higher end of the
rating scale, indicating generally positive customer satisfaction.
However, lower ratings can be investigated to identify operational
problems such as delayed deliveries or poor service.
""")


# ============================================================
# 13. BOX PLOT - REVENUE BY CITY
# ============================================================

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="City",
    y="Revenue"
)

plt.title("Revenue Distribution Across Cities", fontsize=16, fontweight="bold")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The box plot compares revenue distributions between cities.
It shows the median, spread, and potential extreme revenue values.
Cities with larger spreads have greater variation in business
performance across observations.
""")


# ============================================================
# 14. BOX PLOT - DELIVERY TIME BY WEATHER
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Weather",
    y="Avg_Delivery_Minutes"
)

plt.title(
    "Delivery Time Distribution by Weather",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Weather")
plt.ylabel("Average Delivery Time (Minutes)")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
Delivery times vary according to weather conditions.
Rainy conditions generally show higher delivery times compared
with clearer weather conditions. Weather-related delays can
therefore affect delivery efficiency and customer experience.
""")


# ============================================================
# 15. VIOLIN PLOT - CUSTOMER RATING BY ORDER CHANNEL
# ============================================================

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="Order_Channel",
    y="Customer_Rating"
)

plt.title(
    "Customer Rating Distribution by Order Channel",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Order Channel")
plt.ylabel("Customer Rating")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The violin plot shows the distribution of customer ratings for
different order channels. The shape and width of each violin
indicate where ratings are concentrated and how much they vary
between channels.
""")


# ============================================================
# 16. VIOLIN PLOT - REVENUE BY CUISINE
# ============================================================

plt.figure(figsize=(11, 6))

sns.violinplot(
    data=df,
    x="Cuisine",
    y="Revenue"
)

plt.title(
    "Revenue Distribution by Cuisine",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Cuisine")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The violin plot compares both the distribution and spread of
revenue across cuisines. Some cuisines have higher typical
revenue values, while others show greater variation.
""")


# ============================================================
# 17. COUNT PLOT - ORDERS BY WEATHER
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Weather"
)

plt.title(
    "Number of Order Records by Weather Condition",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Weather")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The count plot shows how frequently each weather condition occurs
in the dataset. This provides context for interpreting other
weather-related performance measures such as delivery time,
orders, and customer ratings.
""")


# ============================================================
# 18. COUNT PLOT - ORDER CHANNEL
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="Order_Channel"
)

plt.title(
    "Distribution of Food Delivery Order Channels",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Order Channel")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The count plot compares the number of observations for each
order channel. It shows the relative representation of App,
Website, and Partner Platform orders in the dataset.
""")


# ============================================================
# 19. BAR CHART - PERFORMANCE BY ORDER CHANNEL
# ============================================================

channel_performance = (
    df.groupby("Order_Channel")
    [["Orders", "Revenue"]]
    .mean()
    .reset_index()
)

channel_long = channel_performance.melt(
    id_vars="Order_Channel",
    value_vars=["Orders", "Revenue"],
    var_name="Metric",
    value_name="Average_Value"
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=channel_long,
    x="Order_Channel",
    y="Average_Value",
    hue="Metric"
)

plt.title(
    "Average Orders and Revenue by Order Channel",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Order Channel")
plt.ylabel("Average Value")
plt.legend(title="Metric")
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The comparison highlights differences in performance between
order channels. The App channel has strong average order and
revenue performance in this dataset, while Partner Platform and
Website channels show different performance levels.
""")


# ============================================================
# 20. BAR CHART - WEATHER AND CUSTOMER RATING
# ============================================================

weather_rating = (
    df.groupby("Weather")["Customer_Rating"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=weather_rating,
    x="Weather",
    y="Customer_Rating"
)

plt.title(
    "Average Customer Rating by Weather Condition",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Weather")
plt.ylabel("Average Customer Rating")
plt.tight_layout()
plt.show()

print("\nINTERPRETATION:")

best_weather = weather_rating.iloc[0]
worst_weather = weather_rating.iloc[-1]

print(
    f"{best_weather['Weather']} weather has the highest average "
    f"customer rating, while {worst_weather['Weather']} has the lowest."
)

print("""
Weather conditions can influence the overall customer experience.
Poor weather, especially rain, may increase delivery times and
contribute to lower customer ratings.
""")


# ============================================================
# 21. CORRELATION HEATMAP
# ============================================================

numeric_columns = [
    "Orders",
    "Average_Order_Value",
    "Revenue",
    "Marketing_Spend",
    "Discounts",
    "Avg_Delivery_Minutes",
    "Customer_Rating",
    "Repeat_Customer_Percent"
]

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title(
    "Correlation Heatmap of Food Delivery Performance Metrics",
    fontsize=16,
    fontweight="bold"
)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
The correlation heatmap summarizes relationships between the
numerical business variables.

Important relationships include:

1. Orders and Revenue have a strong positive relationship.
   More orders generally lead to higher revenue.

2. Delivery time and Customer Rating have a strong negative
   relationship. Longer delivery times are associated with
   lower customer ratings.

3. Marketing Spend and Revenue show a positive relationship,
   although the relationship is weaker than the relationship
   between orders and revenue.

4. Customer Rating and Repeat Customer Percentage can be
   examined to understand customer loyalty patterns.
""")


# ============================================================
# 22. MONTHLY PERFORMANCE ANALYSIS
# ============================================================

monthly = (
    df.groupby("Month")
    [["Orders", "Revenue"]]
    .mean()
    .reindex(month_order)
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly,
    x="Month",
    y="Orders",
    marker="o"
)

plt.title(
    "Average Monthly Orders",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Month")
plt.ylabel("Average Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("""
INTERPRETATION:
Monthly order volumes show seasonal variation. Some months
experience stronger demand than others, allowing the business
to identify potential peak periods for promotional campaigns
and operational planning.
""")


# ============================================================
# 23. FINAL BUSINESS SUMMARY
# ============================================================

# Calculate important statistics
top_city = (
    df.groupby("City")["Revenue"]
    .sum()
    .idxmax()
)

top_cuisine = (
    df.groupby("Cuisine")["Revenue"]
    .sum()
    .idxmax()
)

top_order_cuisine = (
    df.groupby("Cuisine")["Orders"]
    .sum()
    .idxmax()
)

best_channel = (
    df.groupby("Order_Channel")["Revenue"]
    .mean()
    .idxmax()
)

weather_delivery = (
    df.groupby("Weather")["Avg_Delivery_Minutes"]
    .mean()
    .sort_values(ascending=False)
)

best_rating_weather = (
    df.groupby("Weather")["Customer_Rating"]
    .mean()
    .idxmax()
)

worst_rating_weather = (
    df.groupby("Weather")["Customer_Rating"]
    .mean()
    .idxmin()
)

print("\n")
print("=" * 70)
print("FINAL SUMMARY OF IMPORTANT FINDINGS")
print("=" * 70)

print(f"""
1. CITY PERFORMANCE
   {top_city} is the highest-revenue city in the dataset.

2. CUISINE PERFORMANCE
   {top_cuisine} generates the highest total revenue.
   {top_order_cuisine} records the highest total number of orders.

3. ORDER CHANNEL
   {best_channel} has the highest average revenue among the
   available order channels.

4. MARKETING AND REVENUE
   Marketing spend has a positive relationship with revenue,
   indicating that higher marketing investment is associated
   with improved revenue in the observed data.

5. DELIVERY TIME AND CUSTOMER SATISFACTION
   Delivery time has a strong negative relationship with
   customer rating. Faster delivery is associated with better
   customer satisfaction.

6. WEATHER IMPACT
   Rainy weather produces the highest average delivery time.
   {best_rating_weather} weather has the highest average
   customer rating, while {worst_rating_weather} has the lowest.

7. ORDERS AND REVENUE
   Orders and revenue have a strong positive relationship,
   meaning higher order volumes generally correspond to higher
   revenue.

8. BUSINESS RECOMMENDATIONS
   - Improve delivery speed, particularly during rainy weather.
   - Invest in high-performing cities and cuisines.
   - Optimize marketing spending based on revenue performance.
   - Strengthen the App channel because of its strong performance.
   - Investigate low-performing cities, cuisines, and channels.
   - Use seasonal demand patterns for staffing and promotions.

Overall, the visualization analysis shows that revenue is strongly
connected with order volume, while customer satisfaction is strongly
affected by delivery time. Weather conditions, city, cuisine, and
order channel also create meaningful differences in business
performance.
""")

print("=" * 70)
print("VISUALIZATION PORTFOLIO COMPLETED SUCCESSFULLY")
print("=" * 70)
```
