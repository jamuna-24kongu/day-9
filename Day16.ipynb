```python
# ============================================================
# DAY 16 - STUDENT WELLBEING STATISTICAL ANALYSIS & PROBABILITY
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: f'{x:.4f}')


# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

# Google Colab:
# Upload the CSV file first using the Files panel.

df = pd.read_csv('/content/Day16_Student_Wellbeing_Survey (1).csv')

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())


# ------------------------------------------------------------
# 3. DATASET INSPECTION
# ------------------------------------------------------------

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Duplicate Rows:")
print(df.duplicated().sum())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDescriptive Statistics:")
display(df.describe())


# ============================================================
# 4. DESCRIPTIVE STATISTICS
# ============================================================

variables = [
    'Weekly_Study_Hours',
    'Average_Sleep_Hours',
    'Daily_Screen_Time_Hours',
    'Stress_Score',
    'Academic_Readiness_Score'
]

results = []

for col in variables:

    data = df[col].dropna()

    mean = data.mean()
    median = data.median()
    mode = data.mode().tolist()

    minimum = data.min()
    maximum = data.max()

    range_value = maximum - minimum

    variance = data.var()
    std_dev = data.std()

    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)

    iqr = q3 - q1

    results.append({
        'Variable': col,
        'Mean': mean,
        'Median': median,
        'Mode': mode,
        'Range': range_value,
        'Variance': variance,
        'Standard Deviation': std_dev,
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr
    })

stats_df = pd.DataFrame(results)

print("\nSTATISTICAL SUMMARY")
display(stats_df)


# ------------------------------------------------------------
# Greatest variability
# ------------------------------------------------------------

greatest_variability = stats_df.loc[
    stats_df['Standard Deviation'].idxmax(),
    'Variable'
]

print("Variable with Greatest Variability:",
      greatest_variability)


# ============================================================
# 5. OUTLIER DETECTION USING IQR METHOD
# ============================================================

outlier_columns = [
    'Weekly_Study_Hours',
    'Daily_Screen_Time_Hours',
    'Commute_Time_Minutes',
    'Monthly_Discretionary_Spending'
]

outlier_results = []

for col in outlier_columns:

    data = df[col].dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data < lower_bound) |
        (data > upper_bound)
    ]

    outlier_results.append({
        'Variable': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Number of Outliers': len(outliers)
    })

    print("\n----------------------------------")
    print(col)
    print("----------------------------------")

    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)

    print("Number of Outliers:", len(outliers))

    print("Outlier Values:")
    print(outliers.tolist())


outlier_df = pd.DataFrame(outlier_results)

print("\nOUTLIER SUMMARY")
display(outlier_df)


# ============================================================
# 6. BEFORE AND AFTER OUTLIER REMOVAL
# ============================================================

# We will use Weekly_Study_Hours

column = 'Weekly_Study_Hours'

data = df[column].dropna()

Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

clean_data = data[
    (data >= lower_bound) &
    (data <= upper_bound)
]

comparison = pd.DataFrame({
    'Mean': [
        data.mean(),
        clean_data.mean()
    ],
    'Median': [
        data.median(),
        clean_data.median()
    ],
    'Count': [
        len(data),
        len(clean_data)
    ]
}, index=[
    'Before Removing Outliers',
    'After Removing Outliers'
])

print("\nBEFORE VS AFTER OUTLIER REMOVAL")
display(comparison)

print(
    "\nInterpretation: Removing extreme observations affects "
    "the mean more strongly than the median because the mean "
    "is sensitive to outliers."
)


# ============================================================
# 7. PROBABILITY EVENTS
# ============================================================

# Event A:
# Student has a Part-Time Job = Yes

A = df['Part_Time_Job'] == 'Yes'


# Event B:
# Student has Stress Score >= 7

B = df['Stress_Score'] >= 7


# Event C:
# Student has Scholarship = Yes

C = df['Scholarship'] == 'Yes'


# Event D:
# Student exercises at least 3 days per week

D = df['Exercise_Days_Per_Week'] >= 3


N = len(df)


# ============================================================
# 8. BASIC PROBABILITIES
# ============================================================

P_A = A.mean()

P_B = B.mean()

P_C = C.mean()

P_D = D.mean()


# P(A or B)

P_A_or_B = (A | B).mean()


# P(A and B)

P_A_and_B = (A & B).mean()


# P(A | B)

P_A_given_B = (A & B).sum() / B.sum()


# P(B | A)

P_B_given_A = (A & B).sum() / A.sum()


probability_results = pd.DataFrame({
    'Probability': [
        P_A,
        P_B,
        P_C,
        P_D,
        P_A_or_B,
        P_A_and_B,
        P_A_given_B,
        P_B_given_A
    ]
}, index=[
    'P(A)',
    'P(B)',
    'P(C)',
    'P(D)',
    'P(A or B)',
    'P(A and B)',
    'P(A | B)',
    'P(B | A)'
])

print("\nPROBABILITY RESULTS")
display(probability_results)


# ============================================================
# 9. PROBABILITY FORMULAS AND INTERPRETATIONS
# ============================================================

print("\nFORMULAS")
print("----------------------------------------")

print("P(A) = Number of A / Total Students")
print("P(B) = Number of B / Total Students")
print("P(C) = Number of C / Total Students")
print("P(D) = Number of D / Total Students")

print("\nP(A or B) = P(A) + P(B) - P(A and B)")

print("\nP(A and B) = Number of students satisfying both A and B / Total")

print("\nP(A | B) = P(A and B) / P(B)")

print("\nP(B | A) = P(A and B) / P(A)")


print("\nINTERPRETATION")
print("----------------------------------------")

print(
    f"P(A) = {P_A:.4f} → "
    f"{P_A*100:.2f}% of students have a part-time job."
)

print(
    f"P(B) = {P_B:.4f} → "
    f"{P_B*100:.2f}% of students have Stress Score >= 7."
)

print(
    f"P(C) = {P_C:.4f} → "
    f"{P_C*100:.2f}% of students have scholarships."
)

print(
    f"P(D) = {P_D:.4f} → "
    f"{P_D*100:.2f}% exercise at least 3 days per week."
)


# ============================================================
# 10. MUTUALLY EXCLUSIVE EVENTS
# ============================================================

Year1 = df['Year_of_Study'] == 1

Year4 = df['Year_of_Study'] == 4

intersection = (Year1 & Year4).sum()

print("\nMUTUALLY EXCLUSIVE TEST")
print("----------------------------------------")

print("Year 1 students:", Year1.sum())
print("Year 4 students:", Year4.sum())

print("Year 1 AND Year 4:", intersection)

if intersection == 0:
    print(
        "Conclusion: Year 1 and Year 4 are mutually exclusive "
        "because a student cannot belong to both years."
    )
else:
    print(
        "Conclusion: Year 1 and Year 4 are not mutually exclusive."
    )


# ============================================================
# 11. TEST WHETHER A AND B ARE INDEPENDENT
# ============================================================

P_A_times_P_B = P_A * P_B

print("\nINDEPENDENCE TEST")
print("----------------------------------------")

print("P(A and B) =", P_A_and_B)

print("P(A) × P(B) =", P_A_times_P_B)

difference = abs(P_A_and_B - P_A_times_P_B)

print("Difference =", difference)

if np.isclose(P_A_and_B, P_A_times_P_B):
    print(
        "Conclusion: A and B appear to be independent."
    )
else:
    print(
        "Conclusion: A and B do NOT appear to be independent."
    )


print(
    "\nInterpretation: For independent events, "
    "P(A and B) should be approximately equal to "
    "P(A) × P(B). Since the two values differ, "
    "A and B appear to be dependent."
)


# ============================================================
# 12. BAYES' THEOREM
# ============================================================

# P(not A)

P_not_A = 1 - P_A


# P(B | not A)

P_B_given_not_A = ((~A) & B).sum() / (~A).sum()


# Bayes Formula:
#
# P(A|B) =
# [P(B|A) * P(A)] /
# [P(B|A)*P(A) + P(B|not A)*P(not A)]

P_A_given_B_Bayes = (
    P_B_given_A * P_A
) / (
    P_B_given_A * P_A +
    P_B_given_not_A * P_not_A
)


print("\nBAYES' THEOREM")
print("----------------------------------------")

print("P(A) =", P_A)

print("P(B | A) =", P_B_given_A)

print("P(B | not A) =", P_B_given_not_A)

print("P(not A) =", P_not_A)

print("\nBayes Formula:")
print(
    "P(A|B) = [P(B|A) × P(A)] / "
    "[P(B|A)×P(A) + P(B|not A)×P(not A)]"
)

print(
    "\nP(A | B) using Bayes =",
    P_A_given_B_Bayes
)

print(
    "P(A | B) using direct calculation =",
    P_A_given_B
)

print(
    "\nDifference =",
    abs(P_A_given_B_Bayes - P_A_given_B)
)

if np.isclose(P_A_given_B_Bayes, P_A_given_B):
    print(
        "\nConclusion: Bayes' theorem is verified because "
        "the Bayes result agrees with the direct conditional "
        "probability calculation."
    )
else:
    print(
        "\nConclusion: The values do not agree. "
        "Check the calculations."
    )


# ============================================================
# 13. NORMAL DISTRIBUTION
# ============================================================

score = df['Academic_Readiness_Score'].dropna()

mean_score = score.mean()

std_score = score.std()


print("\nNORMAL DISTRIBUTION")
print("----------------------------------------")

print("Mean Academic Readiness Score:",
      mean_score)

print("Standard Deviation:",
      std_score)


# ============================================================
# 14. HIGHEST AND LOWEST Z-SCORES
# ============================================================

highest_score = score.max()

lowest_score = score.min()


z_highest = (
    highest_score - mean_score
) / std_score


z_lowest = (
    lowest_score - mean_score
) / std_score


print("\nHIGHEST SCORE")
print("----------------------------------------")

print("Highest Score:", highest_score)

print("Z-score:",
      z_highest)


print("\nLOWEST SCORE")
print("----------------------------------------")

print("Lowest Score:", lowest_score)

print("Z-score:",
      z_lowest)


# ============================================================
# 15. Z-SCORE INTERPRETATION
# ============================================================

print("\nZ-SCORE INTERPRETATION")
print("----------------------------------------")

print(
    f"The highest score has a Z-score of {z_highest:.4f}. "
    f"This means it is approximately "
    f"{abs(z_highest):.2f} standard deviations above "
    f"the mean."
)

print(
    f"The lowest score has a Z-score of {z_lowest:.4f}. "
    f"This means it is approximately "
    f"{abs(z_lowest):.2f} standard deviations below "
    f"the mean."
)


# ============================================================
# 16. 68-95-99.7 EMPIRICAL RULE
# ============================================================

print("\n68-95-99.7 EMPIRICAL RULE")
print("----------------------------------------")

print(
    "Within 1 standard deviation:",
    f"{mean_score - std_score:.2f} to "
    f"{mean_score + std_score:.2f}"
)

print(
    "Expected percentage: 68%"
)


print(
    "\nWithin 2 standard deviations:",
    f"{mean_score - 2*std_score:.2f} to "
    f"{mean_score + 2*std_score:.2f}"
)

print(
    "Expected percentage: 95%"
)


print(
    "\nWithin 3 standard deviations:",
    f"{mean_score - 3*std_score:.2f} to "
    f"{mean_score + 3*std_score:.2f}"
)

print(
    "Expected percentage: 99.7%"
)


# ============================================================
# 17. FINAL FIVE STATISTICAL OBSERVATIONS
# ============================================================

print("\n")
print("=" * 70)
print("FINAL 5 STATISTICAL OBSERVATIONS")
print("=" * 70)


print(
    "\n1. Academic Readiness Score shows the greatest "
    "variability among the five main numerical variables "
    "when standard deviation is used as the measure."
)


print(
    "\n2. Weekly Study Hours contains IQR-based outliers. "
    "Removing these extreme observations changes the mean "
    "more noticeably than the median, demonstrating that "
    "the mean is sensitive to extreme values."
)


print(
    "\n3. Daily Screen Time and the other specified variables "
    "also contain observations outside their IQR bounds, "
    "indicating the presence of unusually high or low values."
)


print(
    "\n4. Events A and B are not independent because "
    "P(A and B) differs from P(A) × P(B). Therefore, "
    "having a part-time job and having a high stress score "
    "show an association in this sample."
)


print(
    "\n5. The highest and lowest Academic Readiness Scores "
    "have substantially different Z-scores. The empirical "
    "rule indicates that approximately 68%, 95%, and 99.7% "
    "of observations are expected within 1, 2, and 3 standard "
    "deviations of the mean, respectively, if the distribution "
    "is approximately normal."
)


print("\n")
print("=" * 70)
print("ANALYSIS COMPLETED")
print("=" * 70)
```
